Import dependencies

```python
import math
import duckdb
import pandas as pd
from processing.core import get_bucket
import matplotlib.pyplot as plt
```

Load the dataset

```python
conn = duckdb.connect("../storage/database/leagues/England.ddb", read_only=True)
table_name = "39-2025-goals-l10"
df = conn.execute(f"SELECT * FROM '{table_name}'").df()
df[["home_team", "away_team", "home_goals", "away_goals", "goals_scored_per_match_diff_bucket_last10", "goals_conceded_per_match_diff_bucket_last10"]]
```

Build results

```python
df["result"] = "draw"

df.loc[df["home_goals"] > df["away_goals"], "result"] = "home_win"
df.loc[df["home_goals"] < df["away_goals"], "result"] = "away_win"
```

Bucket distribution chart

```python
bucket_result = pd.crosstab(
    df["goals_scored_per_match_diff_bucket_last10"],
    df["result"]
)

bucket_result.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("Match Results by Goals-Scored-per-Match_Last5 Bucket")
plt.xlabel("Goals scored per Match Diff Bucket")
plt.ylabel("Number of Matches")
plt.xticks(rotation=0)
plt.legend(title="Result")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
bucket_result = pd.crosstab(
    df["goals_conceded_per_match_diff_bucket_last10"],
    df["result"]
)

bucket_result.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("Match Results by Goals-Conceded-per-Match Bucket")
plt.xlabel("Goals conceded per Match Diff Bucket")
plt.ylabel("Number of Matches")
plt.xticks(rotation=0)
plt.legend(title="Result")
plt.grid(axis="y", alpha=0.3)
plt.show()
```
